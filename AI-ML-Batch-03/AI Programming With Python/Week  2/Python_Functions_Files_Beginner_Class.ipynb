# 1. Function

A **function** is a named block of code you can use again.

Without functions, you copy-paste the same 8 lines like a person who never learned "Save as template".

```text
def greet():
    print("Hello, intern.")
```

- `def` = "I am about to name a recipe"
- `greet` = the name
- `()` = "I am ready. Maybe I will take ingredients later."

**Calling** the function is eating the sandwich: `greet()`

If you write `def` but never call it, you only wrote a recipe and stared at it. Python will not be impressed.


````python
def greet():
    print("Hello, intern. The GPU room is that way.")

greet()
greet()  # see? reusable. Your future self says thanks.
````

# 2. Structure of a Function — Parameters

**Parameters** are empty bowls on the recipe.

When you call the function, you put real food in the bowls. Those real values are **arguments**.

```text
def greet(name):
    print("Hello,", name)
```

`name` is the parameter. `"Rafi"` is the argument.

One function. Many people. That is the whole point. Otherwise you would write `greet_rafi`, `greet_mim`, `greet_nayeem` and your keyboard would file a complaint.


````python
def greet(name):
    print("Hello,", name)

greet("Rafi")
greet("Mim")
greet("GPU")  # it will not answer. still valid Python.
````

````python
def student_line(name, roll):
    print(name, "is roll", roll)

student_line("Nila", 17)
student_line("Tanvir", 24)
````

You can have a **default** parameter. If nobody brings that ingredient, Python uses the house special.

```text
def greet(name, mood="fine"):
    print(name, "is", mood)
```


````python
def greet(name, mood="fine"):
    print(name, "is", mood)

greet("Rafi")
greet("Rafi", "one quiz away from greatness")
````

# 3. Return and Docstring

## `print` vs `return`

This is the classic mix-up. Like `=` vs `==`. People survive. People also suffer.

- `print` = show it on the screen. Then it is gone. Like saying the answer out loud in an exam.
- `return` = give the value back to whoever called you. They can save it, add it, lie about it, whatever.

If you only `print` inside the function, the caller gets `None`. `None` is Python for "I have nothing for you, best of luck."


````python
def add_print_only(a, b):
    print(a + b)          # shows 8, returns nothing useful

def add_return(a, b):
    return a + b          # gives 8 back

x = add_print_only(3, 5)
print("What did I catch?", x)

y = add_return(3, 5)
print("What did I catch?", y)
print("Double it:", y * 2)
````

## Docstring

A **docstring** is a triple-quoted string right under `def`. It is a sticky note for humans.

```text
def add(a, b):
    """Return the sum of a and b."""
    return a + b
```

You read it with `help(add)` or `add.__doc__`.

Write one sentence. Do not write a novel. The function is three lines. The docstring should not be longer than the function's will to live.


````python
def check_pass(marks):
    """Return True if marks are 40 or more."""
    return marks >= 40

print(check_pass(73))
print(check_pass(39))
print(check_pass.__doc__)
````

# 4. Positional Arguments

**Positional** means order matters.

First bowl gets the first value. Second bowl gets the second. Python does not "figure out what you meant." It is not your groupmate.

```text
student_card("Rafi", 101, "CSE")
#            name    roll  dept
```

Swap them and someone named `101` will try to enter the GPU room. Security will have questions.


````python
def student_card(name, roll, dept):
    print(name, "| Roll:", roll, "| Dept:", dept)

student_card("Rafi", 101, "CSE")
student_card("Mim", 102, "EEE")

# This is the crime scene. Uncomment if you enjoy chaos.
# student_card(101, "Rafi", "CSE")
````

# 5. Keyword Arguments

**Keyword** means you label the bowls.

```text
student_card(dept="CSE", name="Rafi", roll=101)
```

Order can change. Python reads the names, not the seating chart.

Mix is allowed: positional first, keywords after.

```text
student_card("Rafi", roll=101, dept="CSE")   # ok
# student_card(name="Rafi", 101, "CSE")      # not ok. Python will hiss.
```

Rule of thumb: once you start naming, keep naming.


````python
def model_info(name, layers, lr=0.01):
    """Return a one-line model registry string. Default lr is 0.01."""
    return f"Model: {name}, Layers: {layers}, LR: {lr}"

print(model_info("TinyNet", 3))                         # positional, default lr
print(model_info(layers=5, name="ResNet"))              # keywords, any order
print(model_info("FastNet", 2, lr=0.001))               # override default
print(model_info(name="SlowNet", layers=4, lr=0.1))
````

# 6. Return Type of a Function

Python functions can return anything: `int`, `float`, `str`, `bool`, `list`, even another function if you are feeling powerful.

Python does not force you to declare the type. It looks at what you `return` and shrugs.

Still, *you* should know what you are returning. If the function is called `average`, do not return a poem.


````python
def average(a, b, c):
    """Return the mean of three numbers as float."""
    return (a + b + c) / 3

def is_even(n):
    """Return True if n is even."""
    return n % 2 == 0

def labels():
    """Return a tiny list of class names."""
    return ["Young", "Middle", "Old"]

acc = average(10, 20, 30)
print(acc, type(acc))          # float. / keeps a decimal on purpose.

print(is_even(18), type(is_even(18)))
print(labels(), type(labels()))
````

You can return **more than one value**. Python packs them in a tuple. Then you can unpack.

```text
def min_max(a, b):
    return a, b    # actually a tuple
```


````python
def min_max(a, b):
    """Return the smaller and the larger of two numbers."""
    if a < b:
        return a, b
    return b, a

lo, hi = min_max(9, 3)
print("low:", lo, "high:", hi)
print(type(min_max(9, 3)))   # tuple. two souvenirs in one bag.
````

# 7. Iterator and Generator

## Iterator (the polite queue)

An **iterable** is something you can loop over: a list, a string, a range.

An **iterator** is the object that actually hands you the next item. `for` builds one for you. You can also do it by hand with `iter()` and `next()`.

When items are finished, `next()` screams `StopIteration`. `for` catches that scream and walks away calmly. You, if you call `next()` too many times, will see the scream.


````python
nums = [10, 20, 30]
it = iter(nums)

print(next(it))
print(next(it))
print(next(it))
# print(next(it))  # uncomment to meet StopIteration
````

## Generator (one plate at a time)

A normal function `return`s once and dies.

A **generator** uses `yield`. It pauses, gives you one value, and remembers where it was. Next time you ask, it continues.

This is how mini-batches work in ML: the GPU cannot eat 20,000 images at once. You yield a plate of 32. Then another plate. The last plate can have leftovers. That is fine.


````python
def count_up(n):
    """Yield numbers from 1 to n. One at a time."""
    i = 1
    while i <= n:
        yield i
        i += 1

for x in count_up(4):
    print("plate:", x)
````

````python
def batch_generator(data, batch_size):
    """Yield slices of data. Last batch may be shorter."""
    if batch_size < 1:
        return
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

samples = [1, 2, 3, 4, 5, 6, 7, 8]
for batch in batch_generator(samples, 3):
    print(batch)
````

# 8. Tiny extra: lambda, map, filter

You need this for a practice problem. Keep it small.

- `lambda` = a nameless one-line function. Like a sticky note, not a novel.
- `filter(fn, data)` = keep items where `fn` is True
- `map(fn, data)` = transform every item

They give you a special object. Wrap with `list(...)` if you want to see it with your eyes.


````python
data = [1, 2, 3, 4, 5, 6]

evens = filter(lambda x: x % 2 == 0, data)
squares = list(map(lambda x: x * x, evens))
print(squares)   # [4, 16, 36]

# Same idea, human names, if lambda looks like a spell:
def is_even(x):
    return x % 2 == 0

def square(x):
    return x * x

print(list(map(square, filter(is_even, data))))
````

# 9. File Handling Basics

A file is a notebook on disk. Python can open it, read it, write it, close it.

If you open a file and never close it, it is like borrowing a pen and walking away. Use `with`. `with` closes the file even if you panic.

## Write a file


````python
# Create a tiny file so later cells have something to read.
with open("hello.txt", "w") as f:
    f.write("Hello, World!")

print("hello.txt is on disk. You did not imagine it.")
````

## Read a file

- `"r"` = read (default)
- `"w"` = write (destroys old content, no mercy)
- `"a"` = append (adds at the end, polite)

`f.read()` = whole file as one string  
`f.readlines()` = list of lines


````python
with open("hello.txt", "r") as f:
    content = f.read()

print(content)
print("type:", type(content))
````

## When the file does not exist

If you try to read a ghost file, Python raises `FileNotFoundError`.

That is not rudeness. That is honesty.

For real programs you catch it with `try / except`. There is also:

- `else` = runs if **no** exception happened
- `finally` = runs **always**. Success, fail, meteor, whatever.

`finally` is the friend who still says "okay bye" after the argument.


````python
def demo_missing():
    try:
        with open("no_such_file.txt", "r") as f:
            print(f.read())
    except FileNotFoundError:
        print("Error: File not found.")
    finally:
        print("Operation complete.")

demo_missing()
````

## Solution 1 — Safe File Read with Full Exception Handling

The job: write `safe_read(filename)` using **try**, **except**, **else**, and **finally**.

| Block | When it runs | What we do |
|-------|----------------|------------|
| `try` | always attempt | open the file and `read()` |
| `except` | file is a ghost | print `Error: File not found.` |
| `else` | try did **not** crash | print the content |
| `finally` | always. success, fail, meteor | print `Operation complete.` |

Why print in `else`, not in `try`?

If `open` fails, we never reach `else`. If `open` works, `else` is the victory lap.

`finally` is the friend who still says bye after the argument. File found? Bye. File missing? Still bye.


````python
def safe_read(filename):
    try:
        f = open(filename, "r")
        content = f.read()
        f.close()
    except FileNotFoundError:
        print("Error: File not found.")
    else:
        print(content)
    finally:
        print("Operation complete.")


# Sample 1: missing file
safe_read("missing.txt")

print("---")

# Sample 2: hello.txt was created in the file-handling section above
safe_read("hello.txt")
````

**Walkthrough**

1. `safe_read("missing.txt")`  
   `open` explodes → `except` prints the error → `finally` still prints complete.  
   Output:
   ```text
   Error: File not found.
   Operation complete.
   ```

2. `safe_read("hello.txt")`  
   `open` works, we store the text → skip `except` → `else` prints `Hello, World!` → `finally` again.  
   Output:
   ```text
   Hello, World!
   Operation complete.
   ```

## Solution 2 — filter even numbers, then map squares

Given `[1, 2, 3, 4, 5, 6]`.

- `filter` = bouncer. Only evens get in: `2, 4, 6`
- `map` = costume person. Square each guest: `4, 16, 36`
- `list(...)` = take a photo. Until you do this, you are holding a sleepy object, not a list you can stare at.

Order matters. Filter first, map second.

If you square first, `1` becomes `1` and still looks even-ish later, or you just keep junk. The problem wants **even, then square**.


````python
data = [1, 2, 3, 4, 5, 6]

evens = filter(lambda x: x % 2 == 0, data)       # 2, 4, 6
squares = list(map(lambda x: x * x, evens))      # 4, 16, 36
print(squares)
````

Same thing with boring honest names, if `lambda` still looks like a spell:


````python
data = [1, 2, 3, 4, 5, 6]

def is_even(x):
    return x % 2 == 0

def square(x):
    return x * x

print(list(map(square, filter(is_even, data))))
````
